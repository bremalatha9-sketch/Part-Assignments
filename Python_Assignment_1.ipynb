```python
# Strings(concatenations,slicing and other methods)
#string concatenations
string1="Hello"
name=input("Enter your name: ")
string2=string1+" "+name
print(string2)
string3="welcome to python programming"
result=string2+","+" " +string3
print(result)
```

```python
#string slicing
#print the first character of the string
#
print("Print the first character of the string   : ",result[0])
print("Print the last character of the string    : ",result[-1])
print("Print the first 5 charcters of the string :",result[:5])
print("Print the last 11 characters of the string:",result[-11:])
print("Print the string in reverse               :",result[::-1])
print("Print the python string                   :",result[28:35])
```

```python
#string methods
strM="Python biginner tutorial"
print("The sentence in UPPERCASE             :",strM.upper())
print("The sentence in Lowercase             :",strM.lower())
print("Capitlized in the original sentence   :",strM.capitalize())
print("Total no. of occurence the t character:",strM.count('t'))
print("Replace the occurence                 :",strM.replace("Python","Data analytics"))
```

```python
#tuple creation,modification and Access
first_tuple=(10,20,30)
second_tuple=(30,50,60)
t_combine=first_tuple+second_tuple
print("Combined tuple   :",t_combine)
print("Repeated tuple   :",t_combine*3)
print("3rd elements     :",t_combine[2])
print("First 3 elements :",t_combine[:3])
print("Last 3 elements  :",t_combine[-4:])
```
