```python
#List (Creation, Modification and Access):
#1. List Creation:
print("----List Creation----")
age_list=[25,26,28,29,23]
print("Age list    :",age_list)
name_list=["Aditya","Ryan","Vijay","Jay","Latha"]
print("Name list   :",name_list)

#2. List Operations / Modifications:
print("----List Operations----")


name_list.append("Yazhini")
print("After append               :",name_list)
age_list.insert(2,30)
print("After insert               :",age_list)
name_list.remove("Yazhini")
print("After remove               :",name_list)
age_list.pop()
print("After pop                  :",age_list)
age_list.extend([29,30,26])
print("After extend               :",age_list)
age_list.reverse()
print("Age in Descending order    :",age_list)
print("Maximum age in the list    :", max(age_list))
print("Minimum age in the list    :", min(age_list))
print("Sum of all ages in the list:", sum(age_list))

#3. Accessing List Elements:
print("----Accesing List of Elements----")
print("The first element of name_list                    :", name_list[:1])
print("The Last element of name_list                     :", name_list[-1:])
print("The elements from index 2 to index 4 in name_list:", name_list[2:4])
print("The elements of name_list in reverse order       :", name_list[::-1])





```

```python

# Dictionary (Creation, Modification and Access):
student_marks = {"Aditya":90,"Ryan":95,"Vijay":80,"Jay":85,"Latha":90}
print("Students Marks           :",student_marks)
print("Aditya scored marks      :",student_marks["Aditya"])
print("Ryan scored marks        :",student_marks["Ryan"])
print("Vijay scored marks       :",student_marks["Vijay"])
print("Jay scored marks         :",student_marks["Jay"])
print("Latha scored marks       :",student_marks["Latha"])
student_marks["Janani"]=90
print("After added janani mark  :",student_marks)
student_marks["Latha"]=82
print("Arter modified latha mark:",student_marks)
print("The keys are             :",student_marks.keys())
print("The values are           :",student_marks.values())
print("The items are            :",student_marks.items())


























```

```python
# Sets (Operations):
my_set={"a","e","i","o","u","a","a","i"}
print(my_set)
print("A set does not allow duplicate values.")
print("A set does not have indexes like a list")
my_set[4] = 's'
print(my_set)
print("A set does not have indexes like a list")
# Explanation:
# Sets do not allow duplicate values.
# So repeated 'a' and 'i' are removed automatically.

```

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print("Set1:",set1)
print("Set2:",set2)
print("Union of set1 and set2:",set1.union(set2))
print("Intersection of set1 and set2:",set1.intersection(set2))
print("Difference of set1 and set2:",set1.difference(set2))
```
