```python

# data analysis using NumPy and Pandas.
# NumPy array operations.
# 1. Create a 1D NumPy Array:
import numpy as np
import pandas as pd
temperatures_w1=np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print(f"Array              :{temperatures_w1}")
# 2. Inspection and Properties:
print(f"Shape of the array :{temperatures_w1.shape}")
print(f"Data type          :{temperatures_w1.dtype}")
print(f"Size of the array  :{temperatures_w1.size}")
```

```python
# 3. Array Oerations
temperatures_f=(temperatures_w1 *9/5)+3
print(f"In Celcious     :{temperatures_w1}")
print(f"In Farhenheit   :{temperatures_f}")
# 4.Array Slicing and Indexing:
print(f"The temperatures for the First three days of this week  :{temperatures_w1[2:]} ")
print(f"The temperatures for the last 2 days of this weekend    :{temperatures_w1[-2:]}")
print(f"The temperatures for the middle 3 days of this week     :{temperatures_w1[2:5]}")
```

```python
#5.Create a 2D Array:
import numpy as np
temperatures_w1=np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
temperatures_w2=np.array([19.2,22.5, 21.3, 24.0, 23.5, 22.8, 20.1])
temperatures=np.array([temperatures_w1, temperatures_w2])
print(f"2D array  :\n{temperatures}")
#6.Inspect and Slice the 2D Array:
print(f"Shape of the array           :{temperatures.shape}")
print(f"Dimension of the array       :{temperatures.ndim}")
print(f"Number of the elements       :{temperatures.size}")
print(f"data type of the array       :{temperatures.dtype}")
print(f"The temperatures of the week1:{temperatures[0,:]}")
print(f"The temperatures of te week2 :{temperatures[1,:]}")
print(f"The last two days of the temperatures for both weekend:\n{temperatures[:,-2:]}")
```

```python
#Pandas Series:
import numpy as np
import pandas as pd
#1.Creating Panda Series
marks=pd.Series([95, 92, 89, 85, 80],index=[ 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print (f"Panda series     :\n{marks}")
#2.Indexing and Slicing:
print(f"The Rank1 marks         :{marks["Rank1"]}")
print(f"First 3 Ranks marks     :{marks.loc["Rank1":"Rank3"]}")
print(f"3 rd Ranked student     :{marks.iloc[2]}")
print(f"Rank where the marks >90:{marks[marks>90]}")
#3.Manipulating Series:
marks.loc["Rank1"]=100
print(f"Manipulating the Rank1 is 100 :{marks}")
marks.drop("Rank5",inplace=True)
print(f"Deleting Rank5 r:{marks}")
cgpa=marks/10
print(f"CGPA:\n{cgpa}")

```

```python
#Pandas DataFrame:
# 1. Creating pandas data frame
import pandas as pd
data={ "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
       "ProductCategory":[ 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                       'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
       "Region": ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South',
                'North'],
       "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
                }


df=pd.DataFrame(data)
print(f"Pandas Data Frame   :\n{df}")
#2,Data Exploration:
print(f"Data frame head operations   :\n{df.head()}")
print(f"Data Frame tail operations   :\n{df.tail()}")
print(f"Shape of the datas           :{df.shape}")
print(f"column of the datas          :\n{df.columns}")
print(f"Data types                   :{df.dtypes}")
print(f"Column access                :\n{df[["ProductCategory","Amount"]]}")
print(f"Last three columns           :\n{df.loc[:,df.columns[-3:]]}")
filtered_data=df[(df["Region"]=="North")&(df["Amount"]>200)]
print(f"filtered data                :\n{filtered_data}")
value_c=df.value_counts("ProductCategory")
print(f"Value count of Product Category column :\n{value_c}")
uniquevalues=df["Region"].nunique()
print(f"Unique values of region:{uniquevalues}")
group_by=df.groupby("Region")["Amount"].mean().reset_index()
print(f"The mean amt of each region    :\n{group_by}")
#3.Manipulating the Data Frame
df.iloc[1,3]=165
print(f"Updating amount   :\n{df}")
df["discount"]=df["Amount"]*10/100
print(f"'Discount' by calculating 10% of the 'Amount'.:\n{df}")
df=df.drop(df[df["TransactionID"]==109].index)
print(f"Remove the row with TransactionID 109.:\n{df}")
d=df.drop("discount",axis=1,inplace=True)

print(f" Delete the 'Discount' column from the DataFrame:\n{df}")

```
